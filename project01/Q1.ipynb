```python
from pyspark import SparkContext
import wget

sc = SparkContext.getOrCreate()
```

```python
url = "https://raw.githubusercontent.com/nongaussian/class-2024-datamining/main/quiz/sherlock.txt"
data_path = wget.download(url, "sherlock.txt")
rdd = sc.textFile(data_path)
```

```python
head_counts = (
    rdd
    .flatMap(lambda line: line.split(" "))
    .filter(lambda word: len(word) > 2)
    .map(lambda word: (word.lower()[:2], 1))
    .reduceByKey(lambda a, b: a + b)
)
```

```python
result = (
    head_counts
    .filter(lambda pair: pair[1] < 32)
    .map(lambda pair: ("total", pair[1]))
    .reduceByKey(lambda a, b: a + b)
    .map(lambda pair: pair[1])
    .collect()
)

print(result[0] if result else 0)
```
