```python
from pyspark.sql import SparkSession
import wget

spark = SparkSession.builder.master("local[*]").appName("Q2").getOrCreate()
sc = spark.sparkContext

# Download the input file, following the course slide example.
wget.download(
    "https://raw.githubusercontent.com/nongaussian/class-2024-datamining/main/quiz/int-pair-input.txt",
    "int-pair-input.txt"
)

lines = sc.textFile("int-pair-input.txt")
pairs = lines.map(lambda line: line.strip()[1:-1].split(",")) \
             .map(lambda values: (int(values[0]), int(values[1])))

first_value = pairs.first()[0]
first_values = pairs.map(lambda pair: pair[0])
maximum = first_values.reduce(lambda a, b: a if a > b else b)
minimum = first_values.reduce(lambda a, b: a if a < b else b)
count_1 = pairs.filter(lambda pair: pair[0] == 1).count()
count_2 = pairs.filter(lambda pair: pair[0] == 2).count()

answer = first_value + maximum + minimum + count_1 + count_2
print("First pair's first element:", first_value)
print("Maximum first element:", maximum)
print("Minimum first element:", minimum)
print("Count of pairs whose first element is 1:", count_1)
print("Count of pairs whose first element is 2:", count_2)
print("Sum:", answer)
```
